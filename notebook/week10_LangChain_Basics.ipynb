```python
from langchain_core.prompts import ChatPromptTemplate

prompts = {
    "Admission": "You are a college admission assistant. Answer this question clearly: {question}",
    "Course": "You are a course information assistant. Answer this question clearly: {question}",
    "Examination": "You are an examination assistant. Answer this question clearly: {question}",
    "Library": "You are a library assistant. Answer this question clearly: {question}",
    "Activities": "You are a college activities assistant. Answer this question clearly: {question}"
}

prompt_templates = {
    category: ChatPromptTemplate.from_template(template)
    for category, template in prompts.items()
}

print("Five prompt templates created successfully!")
```

```python
import os
import pandas as pd

os.makedirs("dataset", exist_ok=True)

prompt_df = pd.DataFrame(
    list(prompts.items()),
    columns=["Category", "Prompt"]
)

prompt_df.to_csv("dataset/prompts.csv", index=False)

print("Prompt templates saved successfully!")
display(prompt_df)
```

```python
chains = {
    category: prompt_templates[category] | llm
    for category in prompts
}

print("Five chatbot chains created successfully!")
```

```python
test_questions = {
    "Admission": "What are the admission requirements?",
    "Course": "What courses are available?",
    "Examination": "How can I prepare for examinations?",
    "Library": "What are the library opening hours?",
    "Activities": "What extracurricular activities are available?"
}

for category, question in test_questions.items():
    print(f"\n--- {category} ---")
    result = chains[category].invoke({"question": question})
    print(result.content)
```

```python
query_records = []
response_records = []

for category, question in test_questions.items():
    result = chains[category].invoke({"question": question})

    query_records.append({
        "Category": category,
        "Question": question
    })

    response_records.append({
        "Category": category,
        "Response": result.content
    })

pd.DataFrame(query_records).to_csv(
    "dataset/user_queries.csv", index=False
)

pd.DataFrame(response_records).to_csv(
    "dataset/chatbot_responses.csv", index=False
)

print("User queries and chatbot responses saved successfully!")
```

```python
from langchain_core.prompts import ChatPromptTemplate

# Summarization chain
summary_prompt = ChatPromptTemplate.from_template(
    "Summarize the following college information in 3 short points:\n{text}"
)

summary_chain = summary_prompt | llm

# Question-answering chain
qa_prompt = ChatPromptTemplate.from_template(
    "Answer the question using the given context.\n"
    "Context: {context}\n"
    "Question: {question}"
)

qa_chain = qa_prompt | llm

print("Summarization chain created successfully!")
print("Question-answering chain created successfully!")
```

```python
college_info = """
Our college offers undergraduate and postgraduate courses.
Students can apply for admission online.
The library provides books, journals, and digital resources.
The college also organizes sports, cultural events, and technical activities.
"""

summary_result = summary_chain.invoke({"text": college_info})

print("Summarized College Information:")
print(summary_result.content)
```

```python
context = """
Our college offers courses in Computer Science, Commerce, and Mathematics.
Students can access books, journals, and digital resources in the library.
Admissions are conducted through the online application process.
"""

question = "What courses are offered by the college?"

qa_result = qa_chain.invoke({
    "context": context,
    "question": question
})

print("Question:", question)
print("\nAnswer:", qa_result.content)
```
